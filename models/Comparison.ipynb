```python
# 1. Vanilla Autoencoder

class VanillaAE(nn.Module):
    def __init__(self, latent_dim=64):
        super().__init__()

        self.encoder = nn.Sequential(
            nn.Flatten(),
            nn.Linear(784, 256),
            nn.ReLU(),
            nn.Linear(256, latent_dim)
        )

        self.decoder = nn.Sequential(
            nn.Linear(latent_dim, 256),
            nn.ReLU(),
            nn.Linear(256, 784),
            nn.Sigmoid()
        )

    def forward(self, x):
        z = self.encoder(x)
        reconstructed = self.decoder(z)

        return reconstructed.view(-1, 1, 28, 28)


# 2. Sparse Autoencoder

class SparseAE(nn.Module):
    def __init__(self, latent_dim=64):
        super().__init__()

        self.encoder = nn.Sequential(
            nn.Flatten(),
            nn.Linear(784, 256),
            nn.ReLU(),
            nn.Linear(256, latent_dim),
            nn.ReLU()
        )

        self.decoder = nn.Sequential(
            nn.Linear(latent_dim, 256),
            nn.ReLU(),
            nn.Linear(256, 784),
            nn.Sigmoid()
        )

    def forward(self, x):
        z = self.encoder(x)
        reconstructed = self.decoder(z)

        return reconstructed.view(-1, 1, 28, 28), z


# 3. Variational Autoencoder

class VAE(nn.Module):
    def __init__(self, latent_dim=64):
        super().__init__()

        self.encoder = nn.Sequential(
            nn.Flatten(),
            nn.Linear(784, 256),
            nn.ReLU()
        )

        self.fc_mu = nn.Linear(256, latent_dim)
        self.fc_logvar = nn.Linear(256, latent_dim)

        self.decoder = nn.Sequential(
            nn.Linear(latent_dim, 256),
            nn.ReLU(),
            nn.Linear(256, 784),
            nn.Sigmoid()
        )

    def reparameterize(self, mu, logvar):
        std = torch.exp(0.5 * logvar)
        eps = torch.randn_like(std)
        return mu + eps * std

    def forward(self, x):
        hidden = self.encoder(x)

        mu = self.fc_mu(hidden)
        logvar = self.fc_logvar(hidden)

        z = self.reparameterize(mu, logvar)

        reconstructed = self.decoder(z)
        reconstructed = reconstructed.view(-1, 1, 28, 28)

        return reconstructed, mu, logvar


# 4. Vanilla GAN

class Generator(nn.Module):
    def __init__(self, latent_dim=100):
        super().__init__()

        self.model = nn.Sequential(
            nn.Linear(latent_dim, 256),
            nn.ReLU(),

            nn.Linear(256, 512),
            nn.ReLU(),

            nn.Linear(512, 784),
            nn.Sigmoid()
        )

    def forward(self, z):
        generated = self.model(z)
        return generated.view(-1, 1, 28, 28)


class Discriminator(nn.Module):
    def __init__(self):
        super().__init__()

        self.model = nn.Sequential(
            nn.Flatten(),

            nn.Linear(784, 512),
            nn.LeakyReLU(0.2),

            nn.Linear(512, 256),
            nn.LeakyReLU(0.2),

            nn.Linear(256, 1)
        )

    def forward(self, x):
        return self.model(x)


# 5. DCGAN

class DCGenerator(nn.Module):
    def __init__(self, latent_dim=100):
        super().__init__()

        self.latent_dim = latent_dim

        self.model = nn.Sequential(
            nn.ConvTranspose2d(
                latent_dim, 128,
                kernel_size=7,
                stride=1,
                padding=0,
                bias=False
            ),
            nn.BatchNorm2d(128),
            nn.ReLU(True),

            nn.ConvTranspose2d(
                128, 64,
                kernel_size=4,
                stride=2,
                padding=1,
                bias=False
            ),
            nn.BatchNorm2d(64),
            nn.ReLU(True),

            nn.ConvTranspose2d(
                64, 1,
                kernel_size=4,
                stride=2,
                padding=1
            ),
            nn.Sigmoid()
        )

    def forward(self, z):
        z = z.view(-1, self.latent_dim, 1, 1)
        return self.model(z)


class DCDiscriminator(nn.Module):
    def __init__(self):
        super().__init__()

        self.model = nn.Sequential(
            nn.Conv2d(
                1, 64,
                kernel_size=4,
                stride=2,
                padding=1,
                bias=False
            ),
            nn.LeakyReLU(0.2, inplace=True),

            nn.Conv2d(
                64, 128,
                kernel_size=4,
                stride=2,
                padding=1,
                bias=False
            ),
            nn.BatchNorm2d(128),
            nn.LeakyReLU(0.2, inplace=True),

            nn.Conv2d(
                128, 1,
                kernel_size=7,
                stride=1,
                padding=0
            )
        )

    def forward(self, x):
        output = self.model(x)
        return output.view(-1, 1)


# 6. DDPM Timestep Embedding

class SinusoidalPosEmb(nn.Module):
    def __init__(self, dim):
        super().__init__()
        self.dim = dim

    def forward(self, t):
        half_dim = self.dim // 2

        emb_factor = math.log(10000) / (half_dim - 1)

        emb = torch.exp(
            torch.arange(
                half_dim,
                device=t.device
            ) * -emb_factor
        )

        emb = t.float()[:, None] * emb[None, :]

        return torch.cat(
            [emb.sin(), emb.cos()],
            dim=1
        )


# DDPM Residual Block

class ResBlock(nn.Module):
    def __init__(
        self,
        in_channels,
        out_channels,
        time_emb_dim
    ):
        super().__init__()

        self.block1 = nn.Sequential(
            nn.Conv2d(
                in_channels,
                out_channels,
                kernel_size=3,
                padding=1
            ),
            nn.GroupNorm(8, out_channels),
            nn.SiLU()
        )

        self.time_mlp = nn.Sequential(
            nn.SiLU(),
            nn.Linear(
                time_emb_dim,
                out_channels
            )
        )

        self.block2 = nn.Sequential(
            nn.Conv2d(
                out_channels,
                out_channels,
                kernel_size=3,
                padding=1
            ),
            nn.GroupNorm(8, out_channels)
        )

        self.residual_conv = (
            nn.Conv2d(
                in_channels,
                out_channels,
                kernel_size=1
            )
            if in_channels != out_channels
            else nn.Identity()
        )

    def forward(self, x, t_emb):
        h = self.block1(x)

        time_emb = self.time_mlp(t_emb)
        h = h + time_emb[:, :, None, None]

        h = self.block2(h)

        return F.silu(
            h + self.residual_conv(x)
        )


# DDPM U-Net

class DDPMUNet(nn.Module):
    def __init__(self, time_emb_dim=128):
        super().__init__()

        self.time_mlp = nn.Sequential(
            SinusoidalPosEmb(time_emb_dim),
            nn.Linear(time_emb_dim, time_emb_dim),
            nn.SiLU(),
            nn.Linear(time_emb_dim, time_emb_dim)
        )

        self.init_conv = nn.Conv2d(
            1, 32, kernel_size=3, padding=1
        )

        # Encoder

        self.down_block1 = ResBlock(
            32, 32, time_emb_dim
        )

        self.downsample1 = nn.Conv2d(
            32, 64,
            kernel_size=4,
            stride=2,
            padding=1
        )

        self.down_block2 = ResBlock(
            64, 64, time_emb_dim
        )

        self.downsample2 = nn.Conv2d(
            64, 128,
            kernel_size=4,
            stride=2,
            padding=1
        )

        self.mid_block = ResBlock(
            128, 128, time_emb_dim
        )

        # Decoder

        self.upsample1 = nn.ConvTranspose2d(
            128, 64,
            kernel_size=4,
            stride=2,
            padding=1
        )

        self.up_block1 = ResBlock(
            128, 64, time_emb_dim
        )

        self.upsample2 = nn.ConvTranspose2d(
            64, 32,
            kernel_size=4,
            stride=2,
            padding=1
        )

        self.up_block2 = ResBlock(
            64, 32, time_emb_dim
        )

        self.output_conv = nn.Conv2d(
            32, 1, kernel_size=1
        )

    def forward(self, x, t):
        t_emb = self.time_mlp(t)

        x = self.init_conv(x)

        # Encode

        h1 = self.down_block1(x, t_emb)
        x = self.downsample1(h1)

        h2 = self.down_block2(x, t_emb)
        x = self.downsample2(h2)

        x = self.mid_block(x, t_emb)

        # Decode + Skip Connections

        x = self.upsample1(x)
        x = torch.cat([x, h2], dim=1)
        x = self.up_block1(x, t_emb)

        x = self.upsample2(x)
        x = torch.cat([x, h1], dim=1)
        x = self.up_block2(x, t_emb)

        return self.output_conv(x)


# 7. MLP Classifier

class MLPClassifier(nn.Module):
    def __init__(self):
        super().__init__()

        self.fc1 = nn.Linear(784, 256)
        self.fc2 = nn.Linear(256, 64)
        self.fc3 = nn.Linear(64, 10)

        self.relu = nn.ReLU()

    def extract_features(self, x):
        x = x.view(x.size(0), -1)

        x = self.relu(self.fc1(x))
        z = self.relu(self.fc2(x))

        return z

    def forward(self, x):
        z = self.extract_features(x)
        return self.fc3(z)


# 8. Frozen MLP Decoder

class FrozenLatentDecoder(nn.Module):
    def __init__(self):
        super().__init__()

        self.decoder = nn.Sequential(
            nn.Linear(64, 256),
            nn.ReLU(),
            nn.Linear(256, 784),
            nn.Sigmoid()
        )

    def forward(self, z):
        reconstructed = self.decoder(z)

        return reconstructed.view(-1, 1, 28, 28)


print("All model architecture classes are ready!")
```

```python
# Load CP

vanilla_ae = VanillaAE().to(DEVICE)
sparse_ae = SparseAE().to(DEVICE)
vae = VAE(latent_dim=LATENT_DIM).to(DEVICE)

generator = Generator(
    GAN_LATENT_DIM
).to(DEVICE)

discriminator = Discriminator().to(DEVICE)

dc_generator = DCGenerator(
    DCGAN_LATENT_DIM
).to(DEVICE)

dc_discriminator = DCDiscriminator().to(DEVICE)

ddpm_model = DDPMUNet(
    time_emb_dim=128
).to(DEVICE)

mlp_classifier = MLPClassifier().to(DEVICE)
mlp_decoder = FrozenLatentDecoder().to(DEVICE)


models = {
    "Vanilla AE": vanilla_ae,
    "Sparse AE": sparse_ae,
    "VAE": vae,
    "Vanilla GAN Generator": generator,
    "Vanilla GAN Discriminator": discriminator,
    "DCGAN Generator": dc_generator,
    "DCGAN Discriminator": dc_discriminator,
    "DDPM": ddpm_model,
    "MLP Classifier": mlp_classifier,
    "Frozen MLP Decoder": mlp_decoder
}

checkpoint_files = {
    "Vanilla AE": "vanilla_ae.pt",
    "Sparse AE": "sparse_ae.pt",
    "VAE": "vae.pt",
    "Vanilla GAN Generator": "vanilla_gan_generator.pt",
    "Vanilla GAN Discriminator": "vanilla_gan_discriminator.pt",
    "DCGAN Generator": "dcgan_generator.pt",
    "DCGAN Discriminator": "dcgan_discriminator.pt",
    "DDPM": "ddpm.pt",
    "MLP Classifier": "mlp_classifier.pt",
    "Frozen MLP Decoder": "mlp_frozen_decoder.pt"
}


for name, model in models.items():

    checkpoint_path = (
        CHECKPOINT_DIR / checkpoint_files[name]
    )

    state_dict = torch.load(
        checkpoint_path,
        map_location="cpu",
        weights_only=True
    )

    model.load_state_dict(
        state_dict,
        strict=True
    )

    model.eval()

    print(f"Loaded: {name}")


prior = torch.load(
    CHECKPOINT_DIR / "mlp_latent_prior.pt",
    map_location="cpu",
    weights_only=True
)

print("\nMLP latent prior keys:", list(prior.keys()))

print("\nAll checkpoints loaded successfully!")
```

```python
# DDPM Schedule and Sampling

betas = torch.linspace(
    0.0001,
    0.02,
    TIMESTEPS,
    device=DEVICE
)

alphas = 1.0 - betas

alphas_cumprod = torch.cumprod(
    alphas,
    dim=0
)

alphas_cumprod_prev = torch.cat([
    torch.ones(1, device=DEVICE),
    alphas_cumprod[:-1]
])

posterior_variance = (
    betas
    * (1.0 - alphas_cumprod_prev)
    / (1.0 - alphas_cumprod)
).clamp(min=1e-20)


def extract(schedule, t, shape):

    values = schedule.gather(0, t)

    return values.reshape(
        t.size(0),
        *([1] * (len(shape) - 1))
    )


# DDPM Reverse Diffusion

@torch.inference_mode()
def p_sample(model, x, t):

    beta_t = extract(
        betas,
        t,
        x.shape
    )

    sqrt_one_minus_alpha_bar = extract(
        torch.sqrt(1.0 - alphas_cumprod),
        t,
        x.shape
    )

    sqrt_recip_alpha = extract(
        torch.sqrt(1.0 / alphas),
        t,
        x.shape
    )

    predicted_noise = model(x, t)

    model_mean = sqrt_recip_alpha * (
        x
        - beta_t
        / sqrt_one_minus_alpha_bar
        * predicted_noise
    )

    noise = torch.randn_like(x)

    nonzero_mask = (
        (t != 0)
        .float()
        .view(-1, 1, 1, 1)
    )

    variance_t = extract(
        posterior_variance,
        t,
        x.shape
    )

    return (
        model_mean
        + nonzero_mask
        * torch.sqrt(variance_t)
        * noise
    )


# DDPM Full Sampling

@torch.inference_mode()
def ddpm_sample(
    model,
    num_samples=16,
    with_trajectory=False
):

    model.eval()

    x = torch.randn(
        num_samples,
        1,
        28,
        28,
        device=DEVICE
    )

    trajectory = (
        {999: x[0].cpu().clone()}
        if with_trajectory
        else {}
    )

    for step in reversed(range(TIMESTEPS)):

        t = torch.full(
            (num_samples,),
            step,
            device=DEVICE,
            dtype=torch.long
        )

        x = p_sample(
            model,
            x,
            t
        )

        if (
            with_trajectory
            and step in [800, 600, 400, 200, 100, 0]
        ):
            trajectory[step] = (
                x[0].cpu().clone()
            )

    return x, trajectory


# VAE Sampling

@torch.inference_mode()
def sample_vae(n):

    z = torch.randn(
        n,
        LATENT_DIM,
        device=DEVICE
    )

    return vae.decoder(z).view(
        n, 1, 28, 28
    )


# Vanilla GAN Sampling

@torch.inference_mode()
def sample_gan(n):

    z = torch.randn(
        n,
        GAN_LATENT_DIM,
        device=DEVICE
    )

    return generator(z)


# DCGAN Sampling

@torch.inference_mode()
def sample_dcgan(n):

    z = torch.randn(
        n,
        DCGAN_LATENT_DIM,
        device=DEVICE
    )

    return dc_generator(z)


# DDPM Sampling

@torch.inference_mode()
def sample_ddpm(n):

    generated, _ = ddpm_sample(
        ddpm_model,
        num_samples=n
    )

    return generated


print("Samplers ready")
```

```python
# Inference Latency Benchmark

BENCHMARK_BATCH_SIZE = 64
DDPM_CHUNK_SIZE = 16
WARMUP = 1
REPEATS = 3



training_times = {}

for name, df in histories.items():

    assert "epoch_time_s" in df.columns, (
        f"Missing epoch_time_s for {name}"
    )

    training_times[name] = float(
        df["epoch_time_s"].mean()
    )

print("Training times loaded successfully!")

benchmark_images = torch.rand(
    BENCHMARK_BATCH_SIZE,
    1, 28, 28,
    device=DEVICE
)

vae_noise = torch.randn(
    BENCHMARK_BATCH_SIZE,
    LATENT_DIM,
    device=DEVICE
)

gan_noise = torch.randn(
    BENCHMARK_BATCH_SIZE,
    GAN_LATENT_DIM,
    device=DEVICE
)

dcgan_noise = torch.randn(
    BENCHMARK_BATCH_SIZE,
    DCGAN_LATENT_DIM,
    device=DEVICE
)

for model in [
    vanilla_ae,
    sparse_ae,
    vae,
    generator,
    dc_generator,
    ddpm_model
]:
    model.eval()

@torch.inference_mode()
def generate_ddpm_batch64():

    generated_chunks = []

    for start in range(
        0,
        BENCHMARK_BATCH_SIZE,
        DDPM_CHUNK_SIZE
    ):

        current_batch = min(
            DDPM_CHUNK_SIZE,
            BENCHMARK_BATCH_SIZE - start
        )

        generated, _ = ddpm_sample(
            ddpm_model,
            num_samples=current_batch,
            with_trajectory=False
        )

        generated_chunks.append(generated)

    return torch.cat(
        generated_chunks,
        dim=0
    )

@torch.inference_mode()
def benchmark_vae():

    return vae.decoder(
        vae_noise
    ).view(-1, 1, 28, 28)


benchmarks = {
    "Vanilla AE": lambda: vanilla_ae(
        benchmark_images
    ),

    "Sparse AE": lambda: sparse_ae(
        benchmark_images
    )[0],

    "VAE": benchmark_vae,

    "Vanilla GAN": lambda: generator(
        gan_noise
    ),

    "DCGAN": lambda: dc_generator(
        dcgan_noise
    ),

    "DDPM": generate_ddpm_batch64
}

def synchronize_device():

    if DEVICE.type == "cuda":
        torch.cuda.synchronize()


benchmark_rows = []
raw_measurements = []

for name, function in benchmarks.items():

    print(f"\nBenchmarking: {name}", flush=True)


    for _ in range(WARMUP):

        with torch.inference_mode():
            warmup_output = function()

        synchronize_device()

    del warmup_output

    latencies = []

    for repeat in range(REPEATS):

        synchronize_device()

        start_time = time.perf_counter()

        with torch.inference_mode():
            output = function()

        synchronize_device()

        elapsed_ms = (
            time.perf_counter() - start_time
        ) * 1000

        assert output.shape == (
            BENCHMARK_BATCH_SIZE, 1, 28, 28
        ), f"Unexpected output shape for {name}"

        assert torch.isfinite(output).all().item(), (
            f"Non-finite output from {name}"
        )

        latencies.append(elapsed_ms)

        raw_measurements.append({
            "Model": name,
            "repeat": repeat + 1,
            "latency_ms": elapsed_ms
        })

        print(
            f"  Repeat {repeat+1}/{REPEATS}: "
            f"{elapsed_ms:.2f} ms",
            flush=True
        )

        del output

    median_latency = float(
        np.median(latencies)
    )

    mean_latency = float(
        np.mean(latencies)
    )

    benchmark_rows.append({
        "Model": name,
        "Training Time (s/epoch)": training_times[name],
        "Inference Latency (ms/batch64)": median_latency,
        "Mean Latency (ms/batch64)": mean_latency
    })

    print(
        f"  Median Latency: {median_latency:.2f} ms",
        flush=True
    )

    pd.DataFrame(
        benchmark_rows
    ).to_csv(
        OUTPUT_DIR / "benchmark_results_final.csv",
        index=False
    )

    pd.DataFrame(
        raw_measurements
    ).to_csv(
        OUTPUT_DIR / "benchmark_raw_measurements.csv",
        index=False
    )


benchmark_results = pd.DataFrame(
    benchmark_rows
)

print("\nInference Latency Benchmark Complete!")

display(benchmark_results)

print(
    "Saved results to:",
    OUTPUT_DIR
)
```

```python
# FID Evaluation

try:
    from torchmetrics.image.fid import FrechetInceptionDistance

except (ImportError, ModuleNotFoundError):
    import subprocess
    import sys

    subprocess.check_call([
        sys.executable,
        "-m",
        "pip",
        "install",
        "-q",
        "torchmetrics[image]",
        "torch-fidelity"
    ])

    from torchmetrics.image.fid import FrechetInceptionDistance


FID_SAMPLES = 1000
FID_BATCH_SIZE = 32
DDPM_FID_BATCH_SIZE = 8

FID_CSV = OUTPUT_DIR / "fid_scores.csv"


assert betas.numel() == TIMESTEPS, (
    "DDPM beta schedule has incorrect size"
)

assert alphas.numel() == TIMESTEPS, (
    "DDPM alpha schedule was overwritten. "
    "Run DDPM Schedule and Sampling again before continuing."
)

assert alphas_cumprod.numel() == TIMESTEPS

print("DDPM schedule verified:", TIMESTEPS)

def prepare_fid_images(images):

    images = images.detach().clamp(0, 1)

    images = (
        images * 255
    ).round().to(torch.uint8)

    return images.repeat(1, 3, 1, 1)

fid_metric = FrechetInceptionDistance(
    feature=2048,
    normalize=False,
    reset_real_features=False
).to(DEVICE)

fid_metric.set_dtype(torch.float64)

print("FID metric initialized!")

print("\nExtracting real Test Set features...")

real_count = 0

with torch.inference_mode():

    for images, _ in test_loader:

        remaining = FID_SAMPLES - real_count

        if remaining <= 0:
            break

        images = images[:remaining].to(DEVICE)

        fid_metric.update(
            prepare_fid_images(images),
            real=True
        )

        real_count += images.size(0)

assert real_count == FID_SAMPLES

print(
    f"Real features ready: {real_count} images"
)


vae.eval()
generator.eval()
dc_generator.eval()
ddpm_model.eval()


@torch.inference_mode()
def fid_sample_vae(n):

    z = torch.randn(
        n, LATENT_DIM,
        device=DEVICE
    )

    return vae.decoder(z).view(
        n, 1, 28, 28
    )


@torch.inference_mode()
def fid_sample_gan(n):

    z = torch.randn(
        n, GAN_LATENT_DIM,
        device=DEVICE
    )

    return generator(z)


@torch.inference_mode()
def fid_sample_dcgan(n):

    z = torch.randn(
        n, DCGAN_LATENT_DIM,
        device=DEVICE
    )

    return dc_generator(z)


@torch.inference_mode()
def fid_sample_ddpm(n):

    images, _ = ddpm_sample(
        ddpm_model,
        num_samples=n,
        with_trajectory=False
    )

    return images


models_for_fid = {
    "VAE": (
        fid_sample_vae,
        FID_BATCH_SIZE
    ),

    "Vanilla GAN": (
        fid_sample_gan,
        FID_BATCH_SIZE
    ),

    "DCGAN": (
        fid_sample_dcgan,
        FID_BATCH_SIZE
    ),

    "DDPM": (
        fid_sample_ddpm,
        DDPM_FID_BATCH_SIZE
    )
}


fid_results = []

for model_name, (
    sample_function,
    batch_size
) in models_for_fid.items():

    fid_metric.reset()

    print(
        f"\nEvaluating FID: {model_name}",
        flush=True
    )

    torch.manual_seed(SEED + 123)

    if DEVICE.type == "cuda":
        torch.cuda.manual_seed_all(SEED + 123)

    generated_count = 0

    with torch.inference_mode():

        while generated_count < FID_SAMPLES:

            n = min(
                batch_size,
                FID_SAMPLES - generated_count
            )

            fake_images = sample_function(n)

            assert torch.isfinite(
                fake_images
            ).all().item(), (
                f"{model_name}: NaN or Inf detected"
            )

            fid_metric.update(
                prepare_fid_images(fake_images),
                real=False
            )

            generated_count += n

            if (
                generated_count % 100 < n
                or generated_count == FID_SAMPLES
            ):
                print(
                    f"  Generated: "
                    f"{generated_count}/{FID_SAMPLES}",
                    flush=True
                )

            del fake_images

    score = float(
        fid_metric.compute().item()
    )

    if not np.isfinite(score):
        raise ValueError(
            f"Invalid FID for {model_name}"
        )

    fid_results.append({
        "Model": model_name,
        "FID": score,
        "Real Samples": FID_SAMPLES,
        "Generated Samples": FID_SAMPLES,
        "Feature Dimension": 2048
    })

    pd.DataFrame(
        fid_results
    ).to_csv(
        FID_CSV,
        index=False
    )

    print(
        f"{model_name} FID: {score:.4f}",
        flush=True
    )

    print(
        f"Saved progress to: {FID_CSV}",
        flush=True
    )

fid_df = pd.DataFrame(
    fid_results
)

print("\nFID Evaluation Complete!")

display(
    fid_df.round(4)
)

print("Saved:", FID_CSV)
```
